```python
# ============================================================
# CREDIT CARD FRAUD DETECTION
# FAST END-TO-END PROJECT
# ============================================================

# -----------------------------
# 1. INSTALL REQUIRED LIBRARIES
# -----------------------------
%pip install -q imbalanced-learn xgboost


# -----------------------------
# 2. IMPORTS
# -----------------------------
import warnings
warnings.filterwarnings("ignore")

import gc
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    ConfusionMatrixDisplay,
    precision_recall_curve,
    roc_curve
)

from imblearn.over_sampling import ADASYN
from xgboost import XGBClassifier

print("All libraries imported successfully.")


# ============================================================
# 3. PROBLEM DEFINITION
# ============================================================

print("=" * 70)
print("CREDIT CARD FRAUD DETECTION")
print("=" * 70)

print("""
Objective:
Build a machine learning system that identifies fraudulent
credit card transactions.

Class 0 = Legitimate transaction
Class 1 = Fraudulent transaction

Main challenge:
Severe class imbalance because fraudulent transactions are
much rarer than legitimate transactions.
""")


# ============================================================
# 4. LOAD DATASET
# ============================================================

FILE_PATH = "/content/creditcard.csv"

feature_columns = (
    ["Time"] +
    [f"V{i}" for i in range(1, 29)] +
    ["Amount"]
)

dtype_map = {col: "float32" for col in feature_columns}
dtype_map["Class"] = "int8"

fraud_df = pd.read_csv(
    FILE_PATH,
    dtype=dtype_map
)

print("\nDataset shape:", fraud_df.shape)

print("\nFirst 5 rows:")
display(fraud_df.head())


# ============================================================
# 5. DATA UNDERSTANDING
# ============================================================

print("\nData types:")
print(fraud_df.dtypes)

print("\nTotal missing values:")
print(fraud_df.isnull().sum().sum())

print("\nDuplicate rows:")
print(fraud_df.duplicated().sum())

print("\nClass distribution:")
print(fraud_df["Class"].value_counts())

print("\nClass percentage:")
print(
    (fraud_df["Class"].value_counts(normalize=True) * 100)
    .round(4)
)


# ============================================================
# 6. EXPLORATORY DATA ANALYSIS
# ============================================================

class_counts = fraud_df["Class"].value_counts().sort_index()

legitimate_count = int(class_counts[0])
fraud_count = int(class_counts[1])

fraud_rate = fraud_df["Class"].mean() * 100

print("\nLegitimate transactions:", f"{legitimate_count:,}")
print("Fraudulent transactions :", f"{fraud_count:,}")
print(f"Fraud rate               : {fraud_rate:.4f}%")

plt.figure(figsize=(7, 5))

class_counts.plot(kind="bar")

plt.title("Transaction Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")
plt.xticks(
    [0, 1],
    ["Legitimate", "Fraud"],
    rotation=0
)

plt.tight_layout()
plt.show()


# Basic numerical statistics
print("\nNumerical summary:")
display(fraud_df.describe().T)


# Transaction amount by class
print("\nAverage transaction amount by class:")
print(
    fraud_df.groupby("Class")["Amount"].mean()
)


# ============================================================
# 7. FEATURE / TARGET SEPARATION
# ============================================================

X = fraud_df.drop(columns=["Class"])
y = fraud_df["Class"]

print("\nFeature matrix shape:", X.shape)
print("Target shape:", y.shape)


# ============================================================
# 8. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# First create untouched test set
X_dev, X_test, y_dev, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Split remaining data into train and validation
X_train, X_val, y_train, y_val = train_test_split(
    X_dev,
    y_dev,
    test_size=0.25,
    random_state=42,
    stratify=y_dev
)

print("\nTrain shape     :", X_train.shape)
print("Validation shape:", X_val.shape)
print("Test shape      :", X_test.shape)

print("\nFraud percentages:")
print(f"Train      : {y_train.mean() * 100:.4f}%")
print(f"Validation : {y_val.mean() * 100:.4f}%")
print(f"Test       : {y_test.mean() * 100:.4f}%")


# ============================================================
# 9. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

X_train_scaled = X_train_scaled.astype(np.float32)
X_val_scaled = X_val_scaled.astype(np.float32)
X_test_scaled = X_test_scaled.astype(np.float32)

print("\nFeature scaling completed.")


# ============================================================
# 10. ADASYN
# ============================================================

print("\nBefore ADASYN:")
print(y_train.value_counts())

# Use a partial oversampling ratio to keep computation fast.
# We do NOT force a full 50:50 balance.
adasyn = ADASYN(
    sampling_strategy=0.20,
    n_neighbors=3,
    random_state=42
)

X_train_resampled, y_train_resampled = adasyn.fit_resample(
    X_train_scaled,
    y_train
)

X_train_resampled = np.asarray(
    X_train_resampled,
    dtype=np.float32
)

y_train_resampled = np.asarray(
    y_train_resampled,
    dtype=np.int8
)

print("\nAfter ADASYN:")
print(
    pd.Series(y_train_resampled).value_counts()
)

print(
    "\nResampled training size:",
    f"{len(y_train_resampled):,}"
)

gc.collect()


# ============================================================
# 11. EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model_name,
    model,
    X_eval,
    y_eval,
    threshold=0.5
):
    probabilities = model.predict_proba(X_eval)[:, 1]

    predictions = (
        probabilities >= threshold
    ).astype(int)

    metrics = {
        "Model": model_name,
        "Threshold": threshold,
        "Accuracy": accuracy_score(
            y_eval,
            predictions
        ),
        "Precision": precision_score(
            y_eval,
            predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_eval,
            predictions,
            zero_division=0
        ),
        "F1": f1_score(
            y_eval,
            predictions,
            zero_division=0
        ),
        "ROC_AUC": roc_auc_score(
            y_eval,
            probabilities
        ),
        "PR_AUC": average_precision_score(
            y_eval,
            probabilities
        )
    }

    return metrics, probabilities, predictions


# ============================================================
# 12. LOGISTIC REGRESSION BASELINE
# ============================================================

print("\n" + "=" * 70)
print("MODEL 1: LOGISTIC REGRESSION")
print("=" * 70)

logistic_model = LogisticRegression(
    max_iter=500,
    class_weight="balanced",
    solver="liblinear",
    random_state=42
)

# Use original training data for the baseline
logistic_model.fit(
    X_train_scaled,
    y_train
)

logistic_metrics, logistic_proba, logistic_pred = evaluate_model(
    "Logistic Regression",
    logistic_model,
    X_val_scaled,
    y_val
)

print(logistic_metrics)


# ============================================================
# 13. DECISION TREE
# ============================================================

print("\n" + "=" * 70)
print("MODEL 2: DECISION TREE")
print("=" * 70)

dt_model = DecisionTreeClassifier(
    max_depth=8,
    min_samples_leaf=5,
    class_weight="balanced",
    random_state=42
)

# Class weighting handles imbalance without requiring
# another synthetic dataset for this baseline.
dt_model.fit(
    X_train_scaled,
    y_train
)

dt_metrics, dt_proba, dt_pred = evaluate_model(
    "Decision Tree",
    dt_model,
    X_val_scaled,
    y_val
)

print(dt_metrics)


# ============================================================
# 14. RANDOM FOREST
# ============================================================

print("\n" + "=" * 70)
print("MODEL 3: RANDOM FOREST")
print("=" * 70)

rf_model = RandomForestClassifier(
    n_estimators=60,
    max_depth=10,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train_scaled,
    y_train
)

rf_metrics, rf_proba, rf_pred = evaluate_model(
    "Random Forest",
    rf_model,
    X_val_scaled,
    y_val
)

print(rf_metrics)


# ============================================================
# 15. XGBOOST
# ============================================================

print("\n" + "=" * 70)
print("MODEL 4: XGBOOST")
print("=" * 70)

xgb_model = XGBClassifier(
    n_estimators=120,
    max_depth=4,
    learning_rate=0.10,
    subsample=0.80,
    colsample_bytree=0.80,
    objective="binary:logistic",
    eval_metric="aucpr",
    tree_method="hist",
    random_state=42,
    n_jobs=-1
)

# XGBoost uses the ADASYN-resampled training set
xgb_model.fit(
    X_train_resampled,
    y_train_resampled
)

xgb_metrics, xgb_proba, xgb_pred = evaluate_model(
    "XGBoost",
    xgb_model,
    X_val_scaled,
    y_val
)

print(xgb_metrics)


# ============================================================
# 16. MODEL COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

comparison = pd.DataFrame([
    logistic_metrics,
    dt_metrics,
    rf_metrics,
    xgb_metrics
])

comparison = comparison.sort_values(
    by="PR_AUC",
    ascending=False
).reset_index(drop=True)

display(comparison)


# ============================================================
# 17. XGBOOST VALIDATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("XGBOOST VALIDATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_val,
        xgb_pred,
        target_names=[
            "Legitimate",
            "Fraud"
        ],
        zero_division=0
    )
)


# ============================================================
# 18. XGBOOST CONFUSION MATRIX
# ============================================================

ConfusionMatrixDisplay.from_predictions(
    y_val,
    xgb_pred,
    display_labels=[
        "Legitimate",
        "Fraud"
    ]
)

plt.title(
    "XGBoost Validation Confusion Matrix"
)

plt.tight_layout()
plt.show()


# ============================================================
# 19. THRESHOLD ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("THRESHOLD ANALYSIS")
print("=" * 70)

thresholds = np.arange(
    0.05,
    0.96,
    0.05
)

threshold_results = []

for threshold in thresholds:

    predictions = (
        xgb_proba >= threshold
    ).astype(int)

    threshold_results.append({
        "Threshold": threshold,

        "Precision": precision_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_val,
            predictions,
            zero_division=0
        ),

        "Accuracy": accuracy_score(
            y_val,
            predictions
        )
    })

threshold_df = pd.DataFrame(
    threshold_results
)

display(threshold_df)


# ============================================================
# 20. THRESHOLD PLOT
# ============================================================

plt.figure(figsize=(9, 6))

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    threshold_df["Threshold"],
    threshold_df["Recall"],
    marker="o",
    label="Recall"
)

plt.plot(
    threshold_df["Threshold"],
    threshold_df["F1"],
    marker="o",
    label="F1"
)

plt.xlabel("Classification Threshold")
plt.ylabel("Score")

plt.title(
    "XGBoost Precision / Recall / F1 vs Threshold"
)

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 21. SELECT BEST THRESHOLD
# ============================================================

best_threshold_row = threshold_df.loc[
    threshold_df["F1"].idxmax()
]

best_threshold = float(
    best_threshold_row["Threshold"]
)

print("\nBest threshold based on validation F1:")
display(best_threshold_row)


# ============================================================
# 22. XGBOOST AT BEST THRESHOLD
# ============================================================

best_xgb_pred = (
    xgb_proba >= best_threshold
).astype(int)

print("\nXGBoost performance at selected threshold:")

print(
    classification_report(
        y_val,
        best_xgb_pred,
        target_names=[
            "Legitimate",
            "Fraud"
        ],
        zero_division=0
    )
)


# ============================================================
# 23. ROC CURVE
# ============================================================

fpr, tpr, _ = roc_curve(
    y_val,
    xgb_proba
)

validation_roc_auc = roc_auc_score(
    y_val,
    xgb_proba
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"XGBoost ROC-AUC = {validation_roc_auc:.4f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title(
    "XGBoost Validation ROC Curve"
)

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 24. PRECISION-RECALL CURVE
# ============================================================

precision_curve, recall_curve, _ = precision_recall_curve(
    y_val,
    xgb_proba
)

validation_pr_auc = average_precision_score(
    y_val,
    xgb_proba
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall_curve,
    precision_curve,
    label=f"PR-AUC = {validation_pr_auc:.4f}"
)

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title(
    "XGBoost Validation Precision-Recall Curve"
)

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


# ============================================================
# 25. FEATURE IMPORTANCE
# ============================================================

feature_importance = pd.Series(
    xgb_model.feature_importances_,
    index=X.columns
).sort_values(
    ascending=False
)

print("\nTop 15 important features:")
display(
    feature_importance.head(15)
)

plt.figure(figsize=(10, 6))

feature_importance.head(15).sort_values().plot(
    kind="barh"
)

plt.title(
    "Top XGBoost Feature Importances"
)

plt.xlabel("Importance")

plt.tight_layout()
plt.show()


# ============================================================
# 26. FINAL TRAINING DATA
# ============================================================

print("\n" + "=" * 70)
print("PREPARING FINAL TRAINING DATA")
print("=" * 70)

# Combine training + validation data
X_train_full = pd.concat(
    [X_train, X_val],
    axis=0
)

y_train_full = pd.concat(
    [y_train, y_val],
    axis=0
)

print(
    "Full training data:",
    X_train_full.shape
)

print(
    "Full training labels:",
    y_train_full.shape
)


# ============================================================
# 27. FINAL SCALING
# ============================================================

final_scaler = StandardScaler()

X_train_full_scaled = final_scaler.fit_transform(
    X_train_full
)

X_test_final_scaled = final_scaler.transform(
    X_test
)

X_train_full_scaled = X_train_full_scaled.astype(
    np.float32
)

X_test_final_scaled = X_test_final_scaled.astype(
    np.float32
)

print("Final scaling completed.")


# ============================================================
# 28. FINAL ADASYN
# ============================================================

print("\nApplying final ADASYN...")

final_adasyn = ADASYN(
    sampling_strategy=0.20,
    n_neighbors=3,
    random_state=42
)

X_train_full_resampled, y_train_full_resampled = (
    final_adasyn.fit_resample(
        X_train_full_scaled,
        y_train_full
    )
)

X_train_full_resampled = np.asarray(
    X_train_full_resampled,
    dtype=np.float32
)

y_train_full_resampled = np.asarray(
    y_train_full_resampled,
    dtype=np.int8
)

print(
    "Final resampled training size:",
    f"{len(y_train_full_resampled):,}"
)

gc.collect()


# ============================================================
# 29. FINAL XGBOOST MODEL
# ============================================================

print("\nTraining final XGBoost model...")

final_xgb_model = XGBClassifier(
    n_estimators=120,
    max_depth=4,
    learning_rate=0.10,
    subsample=0.80,
    colsample_bytree=0.80,
    objective="binary:logistic",
    eval_metric="aucpr",
    tree_method="hist",
    random_state=42,
    n_jobs=-1
)

final_xgb_model.fit(
    X_train_full_resampled,
    y_train_full_resampled
)

print("Final XGBoost model trained successfully.")


# ============================================================
# 30. FINAL TEST PREDICTIONS
# ============================================================

final_test_proba = final_xgb_model.predict_proba(
    X_test_final_scaled
)[:, 1]

final_test_pred = (
    final_test_proba >= best_threshold
).astype(int)


# ============================================================
# 31. FINAL TEST METRICS
# ============================================================

final_accuracy = accuracy_score(
    y_test,
    final_test_pred
)

final_precision = precision_score(
    y_test,
    final_test_pred,
    zero_division=0
)

final_recall = recall_score(
    y_test,
    final_test_pred,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    final_test_pred,
    zero_division=0
)

final_roc_auc = roc_auc_score(
    y_test,
    final_test_proba
)

final_pr_auc = average_precision_score(
    y_test,
    final_test_proba
)


# ============================================================
# 32. PRINT FINAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("FINAL TEST RESULTS")
print("=" * 70)

print(f"Accuracy : {final_accuracy:.4f}")
print(f"Precision: {final_precision:.4f}")
print(f"Recall   : {final_recall:.4f}")
print(f"F1-score : {final_f1:.4f}")
print(f"ROC-AUC  : {final_roc_auc:.4f}")
print(f"PR-AUC   : {final_pr_auc:.4f}")
print(f"Threshold: {best_threshold:.2f}")


# ============================================================
# 33. FINAL CLASSIFICATION REPORT
# ============================================================

print("\nFinal classification report:")

print(
    classification_report(
        y_test,
        final_test_pred,
        target_names=[
            "Legitimate",
            "Fraud"
        ],
        zero_division=0
    )
)


# ============================================================
# 34. FINAL CONFUSION MATRIX
# ============================================================

ConfusionMatrixDisplay.from_predictions(
    y_test,
    final_test_pred,
    display_labels=[
        "Legitimate",
        "Fraud"
    ]
)

plt.title(
    "Final XGBoost Test Confusion Matrix"
)

plt.tight_layout()
plt.show()


# ============================================================
# 35. SAVE FINAL RESULTS
# ============================================================

final_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "ROC-AUC",
        "PR-AUC",
        "Threshold"
    ],

    "Value": [
        final_accuracy,
        final_precision,
        final_recall,
        final_f1,
        final_roc_auc,
        final_pr_auc,
        best_threshold
    ]
})

print("\nFinal results table:")
display(final_results)

final_results.to_csv(
    "final_results.csv",
    index=False
)


# ============================================================
# 36. SAVE MODEL
# ============================================================

joblib.dump(
    final_xgb_model,
    "fraud_xgboost_model.pkl"
)

joblib.dump(
    final_scaler,
    "fraud_scaler.pkl"
)

joblib.dump(
    best_threshold,
    "fraud_threshold.pkl"
)

print("\nSaved files:")
print("1. final_results.csv")
print("2. fraud_xgboost_model.pkl")
print("3. fraud_scaler.pkl")
print("4. fraud_threshold.pkl")

print("\n" + "=" * 70)
print("PROJECT COMPLETE")
print("=" * 70)
```
